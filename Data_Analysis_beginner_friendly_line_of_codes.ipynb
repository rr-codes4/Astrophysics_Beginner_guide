Here I have lined up some of the most basic and beginner-friendly data analysis codes which includes pandas,polars,numpy,matplotlib.

```python
import polars as pl
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
print(url)
df = pl.read_csv(url)
#Here we are just using simple algorithms to get a concise summary of the csv
display(df.head()) #The output clearly explains there functions
display(df.shape)
display(df.glimpse())
display(df.describe())
display(df['petal_length'].describe())
#you can display what you want based on values like this in polars and pandas
# It's not possible to generate a visual line graph directly with Polars without importing a dedicated plotting library like matplotlib, which was explicitly excluded.
```

```python
import pandas as pd #we are importing pandas

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
print(f"Attempting to read CSV from: {url}")
pdf = pd.read_csv(url)
#pandas and polars have similiarities
display(pdf.head())
display(pdf.shape)
display(pdf.info)
display(pdf.describe())
display(pdf['petal_length'].describe())
pdf['sepal_length'].plot(kind='box',vert=True,figsize=(14,6))
  #here the axis of the graph can differ depending on what you set vert to True of False

#there are more kind of plots
display(pdf['petal_length'])
```

In numpy the array has normal indices as list but for 2D and 3D one index for one particular bracket.
But in this we are not going to cover slicing and indexing but some common codes that is crucial for data analysis.

```python
import numpy as np
my_list = [1,2,3,4,5,6,7,8] #To use any list in numpy it must be used by passing as an argument to np.array
array = np.array(my_list) #It is not nescessary we can directly pass a list into np.array e.g. np.array([1,2,3])
#The above array is 1D
my_list2 = [[1,2],[3,4],[5,6]]#This is the 2D one
array2 = np.array(my_list2)
my_list3 = [[[1,2],[3,4]],[[5,6],[7,8]]] #This one is 3D
array3 = np.array(my_list3)
array4 = np.zeros((5,6)) #Create an array full of zeroes of 5 rows and 6 columns
array5 = np.random.random() #Create an array of random number between 0 & 1 alse a Tuple
print("Printing the array full of zeroes",array4)
print("Printing the random array",array5)
array6 = np.arange(0,3)#Create a sequenced array of number
print("Sequnced array",array6)

#Statitics using Numpy
print(array.sum(),array2.sum(),array3.sum(axis=0))
print(array.mean(),array2.mean())
print(array.std(),array2.std())
print(array.var())
#if i want to manipulate the values inside the arrays
print("When we will add something directly into an array", [i+100 for i in array])
print("we can also do operation directly in a list",[i*100 for i in my_list] )
```
